# Simple cooling power model

Three loads: sensible cooling of the mass, heat leaking in through the wall, and heat carried in by warm inflow.

```python
import math

# --- inputs ---
specific_heat = 4184.0              # J/(kg·K), water
density = 1000.0                    # kg/m³

cooled_mass = 0.05                  # kg
initial_temperature = 37.0          # °C
target_temperature = 33.0           # °C
outer_temperature = 37.0            # °C, surrounding tissue stays warm
cooling_time = 30 * 60              # seconds

inner_radius = 5e-3                 # m
inner_height = 100e-3               # m
heat_transfer_coefficient = 500.0   # W/(m²·K)

flow_rate = 1.0                     # mL/s, warm fluid entering the cooled zone

# --- geometry ---
wall_area = 2 * math.pi * inner_radius * inner_height

# --- temperature differences ---
cooling_temperature_drop = initial_temperature - target_temperature
mean_inner_temperature = (initial_temperature + target_temperature) / 2
wall_temperature_difference = outer_temperature - mean_inner_temperature

# --- 1) sensible: energy to cool the mass ---
sensible_energy = cooled_mass * specific_heat * cooling_temperature_drop   # J
sensible_power = sensible_energy / cooling_time                            # W

# --- 2) wall: heat leaking in from warm surroundings ---
wall_power = heat_transfer_coefficient * wall_area * wall_temperature_difference  # W
wall_energy = wall_power * cooling_time                                           # J

# --- 3) flow: warm fluid that must be cooled as it enters ---
mass_flow_rate = (flow_rate / 1e6) * density                               # kg/s
flow_power = mass_flow_rate * specific_heat * cooling_temperature_drop     # W
flow_energy = flow_power * cooling_time                                    # J

# --- total ---
total_energy = sensible_energy + wall_energy + flow_energy
total_power = total_energy / cooling_time

print(f"Sensible:  {sensible_power:.2f} W")
print(f"Wall:      {wall_power:.2f} W")
print(f"Flow:      {flow_power:.2f} W")
print(f"Total:     {total_power:.2f} W")
```
